Sketch typical (squared) bias, variance, training error, test error, and Bayes (irreducible) error curves as a function of model flexibility, and explain their shapes.

```python
import numpy as np
from matplotlib.pyplot import subplots
```

```python
flexibility = np.linspace(1, 10, 200)  # x-axis: model flexibility
bias_sq = (10 / flexibility) ** 2       # bias^2 falls as flexibility increases
variance = 0.05 * flexibility ** 2      # variance rises as flexibility increases
bayes_error = np.full_like(flexibility, 2.0)   # irreducible error is constant
train_error = bias_sq * 0.4 + 1.5 / flexibility  # training error decreases monotonically
test_error = bias_sq + variance + bayes_error    # test error = bias^2 + variance + irreducible
```

```python
fig, ax = subplots(figsize=(8, 8))
ax.plot(flexibility, bias_sq, label='(Squared) bias')
ax.plot(flexibility, variance, label='Variance')
ax.plot(flexibility, train_error, label='Training error')
ax.plot(flexibility, test_error, label='Test error')
ax.plot(flexibility, bayes_error, label='Bayes (irreducible) error', linestyle='--')
ax.set_xlabel('Flexibility')
ax.set_ylabel('Error / value')
ax.set_title('Bias-Variance Tradeoff')
ax.legend();
```

**(b) Why each curve has this shape:**

- **Bias²** falls as flexibility grows, since a more flexible method can conform more closely to the true f.
- **Variance** rises as flexibility grows, since a more flexible method fits the training data more closely and is more sensitive to changes in the training set.
- **Training error** decreases monotonically with flexibility, since a more flexible model can better fit the observed data points.
- **Test error** follows a U-shape: it initially falls as bias drops faster than variance rises, then increases once variance dominates (overfitting).
- **Bayes (irreducible) error** stays flat, since it reflects noise in the data (Var(ε)) that no model, however flexible, can remove; it forms a lower bound below which test error cannot fall.
